```python
import numpy as np
print('numpy version:', np.__version__)
```

```python
# single dimensional arrays (Vector)
oneD = np.array( [1, 2, 3])

print(oneD)

# Two dimensional array (Matrix) Grid. Cery to a table
twoD = np.array(
    [
        [1, 2, 3],
        [3, 4, 6],
        [7, 8, 9]
    ]
)

print(twoD)

# Three dimensional Array (Tensor). set of matrices
threeD = np.array(
    [
        [
            [1, 2],
            [3, 4]
        ],
        [
            [5, 6],
            [7, 8]
        ]
    ]
)

print(threeD)
```
